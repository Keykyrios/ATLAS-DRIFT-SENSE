# ATLAS Threshold Calibration

Uses Isotonic Regression to map intermediate stage scores (NCC, HDC, Topo) into a calibrated Confidence measure.

```python
import numpy as np
from sklearn.isotonic import IsotonicRegression

# Placeholder for actual validation data
# In practice, this would load results from eval/runner.py
ncc_scores = np.random.uniform(0.5, 1.0, 100)
errors_px = np.random.uniform(0, 20, 100)

# Labels: 1 if error < 2px else 0
y = (errors_px < 2.0).astype(float)

iso_reg = IsotonicRegression(out_of_bounds='clip')
iso_reg.fit(ncc_scores, y)

print("Calibration complete. Use mapped probabilities as weights in stage6_confidence.py")
```
